```python
with open("../data/putin_complete.json", "r") as f:
    putin_data = pd.read_json(f)

df = putin_data[300:400]

df = df[
    [
        "date",
        "title",
        "place",
        "tags",
        "transcript_filtered"
    ]
]

df.to_csv(
    "speeches_sample.csv",
    index=False,
    encoding="utf-8"
)
```
