```python
import json
from sentence_transformers import SentenceTransformer
import os
from workers.utils.task_definitions import *
from workers.utils.dataset_filters_definitions import *

model = SentenceTransformer(os.getenv("EMBEDDING_MODEL"))
TASK_FILE = "workers/utils/task_definitions/embeddings.json"
```

```python
task_classes = [
    TermDistribution,
    WordCloud,
    RelationFinder,
    RelatedTermCounts,
]
```

```python
tasks = {}
for task_type in task_classes:
    tasks[task_type.name] = [float(x) for x in model.encode(task_type.description)]

with open(TASK_FILE, 'w') as file:
    json.dump(tasks, file)
```
